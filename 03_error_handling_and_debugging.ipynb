#### Error handling and Debugging

###### Error handling:
###### A python mechanism to manage and respond to error (exceptions) that occur during program execution, preventing crashes.

###### Purpose: Ensures programs include

```python
# TypeError handling

try:
    result = '5' + 5
except TypeError:
    print("Caught a TypeError: Cannot concatenate str and int")
```

```python
# FileNotFoundError handling
try:
    with open('non_existent_file.txt', 'r') as file:
        content = file.read()
except FileNotFoundError:
    print("Caught a FileNotFoundError: The file does not exist")
```

```python
# KeyError handling
try:
    my_dict = {'a': 1, 'b': 2}
    value = my_dict['c']
except KeyError:
    print("Caught a KeyError: The key does not exist in the dictionary")
```
